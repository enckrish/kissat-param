```python
from tqdm.notebook import tqdm


internal_nodes = set(aig.nodes).difference(set(aig.input_ids)).difference(set([0]))
for n in tqdm(internal_nodes):
	for fanin_id in [0, 1]:
		goat = aig.nodes[n].fanins[fanin_id]
		aig.nodes[n].fanins[fanin_id] = (goat[0], not goat[1])

		aig.write_cnf_to_file("temp.cnf")
		cadical_check("temp.cnf")
```
